```python
OPENROUTER_API_KEY = os.getenv("OPENRouter_API_KEY")
```

```python
client = OpenAI(
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)
```

```python
# create an empty history
history = [
    {"role": "system", "content": "You are a helpful and friendly AI assistant."}
]

# write a welcome message to user
print("Bot: Salam! Mən sizin AI köməkçinizəm. Çıxış üçün 'exit' və ya 'quit' yazın.\n")

# start flow
while True:
    # get message from user
    user_input = input("Siz: ")
    
    # check message if user wants to quit chat
    if user_input.lower() in ["exit", "quit"]:
        print("Bot: Sağ olun, növbəti görüşlərədək!")
        break
        
    # add user message to history
    history.append({"role": "user", "content": user_input})
    
    # get answer from ai assistant
    response = client.chat.completions.create(
        model="openrouter/auto",
        messages=history
    )
    bot_answer = response.choices[0].message.content
    
    # add answer to history
    history.append({"role": "assistant", "content": bot_answer})
    
    # print answer to screen
    print(f"Bot: {bot_answer}\n")
```

```python
def start_chat(client, model="openrouter/auto"):
    history = [
        {"role": "system", "content": "You are a helpful and friendly AI assistant."}
    ]
    
    print("Bot: Salam! Mən sizin AI köməkçinizəm. Çıxış üçün 'exit' və ya 'quit' yazın.\n")
    
    while True:
        user_input = input("Siz: ")
        
        if user_input.lower() in ["exit", "quit"]:
            print("Bot: Sağ olun, növbəti görüşlərədək!")
            break
            
        history.append({"role": "user", "content": user_input})
        
        response = client.chat.completions.create(
            model=model,
            messages=history
        )
        bot_answer = response.choices[0].message.content
        
        history.append({"role": "assistant", "content": bot_answer})
        print(f"Bot: {bot_answer}\n")

```

```python
start_chat(client)
```
